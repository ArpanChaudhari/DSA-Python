# Decimal to Binary

```python
def binary(n:int)->int:
    strB= ""
    while n != 1:
        if  n % 2 == 1:
            strB += '1'
        else :
            strB += "0"
        n //= 2
    return strB[::-1]
```

```python
n = 14
print(binary(n))
```

# Binary To Decimal

```python
def Decimal(st:str)-> int:
    st = int(st)
    power = 0
    number = 0


    while st > 0 :
        rem = st % 10
        number += (2**power) * rem
        power += 1
        st //= 10

    return number
```

```python
str1 = "1101"
print(Decimal(str1))
```

# Check if i'th bit is set or not

```python
"""
n = 20 => 10100
i = 2 => in n=20 ith(2) bit is 1 so,its set bit
1<<i = 100

then n & (1<<i) 
    10100
&   00100  
  = 00100 => 4


n = 20
i = 3 => in n =20 ith(3) bit is 0 so,its unset bit
1<<i = 1000

then 
    10100
&   01000
=   00000 => 0


from both the above examples we can say that
if n & (1<<i) != 0 => ith bit is set
if n & (1<<i) == 0 => ith bit is unset
"""
```

```python
n = 20 
i = 3 

print(n & (1<<i) != 0)

i= 2
print(n & (1<<i) != 0)
```

# Set The i'th bit

```python
"""
n = 20 => 10100
i = 2 
in n = 20 i'th(2) bit is 1 so,its set bit
but we don't know the value of that bit so we can use the formula 
i<<i => 100

then n | (1<<i)
    10100
|   00100
=   10100 => 20

n = 20 => 10100
i = 1 
i<<i => 010

then n | (1<<i)
    10100
|   00010
=   10110 => 22
    """
```

```python
n = 20 
print(bin(n)[2:])
i = 3
n = n | (1<<i)
print(f"{bin(n)[2:]} => {n}")
```

# Clear the i'th bit

```python
"""
for set we can apply or with ith place 1
for clear we can apply and with ith place 0

for ith place 0 we do ~(1<<i)
n = 20 => 10100
i = 2
1<<i = 100
then ~(1<<i) = 011

then n & ~(1<<i)
    10100
&   11011
=   10000 => 16
n = 20 => 10100
i = 3
1<<i = 1000
then ~(1<<i) = 10111
then n & ~(1<<i)
    10100
&   10111
=   10100 => 20

"""
```

```python
n = 20 
print(bin(n)[2:])
i = 2
n = n & ~(1<<i)
print(bin(n)[2:])
```

```python
n = 20 
print(bin(n)[2:])
i = 2
n = n ^ (1<<i)
print(bin(n)[2:])

n = 20
i = 3
n = n ^ (1<<i)
print(bin(n)[2:])
```

# Remove  the Last set bit (Rightmost)

```python
"""
7 = 111
6 = 110
7 & 6 = 110 => last set bit removed

13 = 1101
12 = 1100
13 & 12 = 1100 => last set bit removed

from above we can say that 
n & (n-1) => remove the last set bit 
"""
```

```python
n = 20
print(bin(n)[2:])
n = n & (n-1)
print(bin(n)[2:])
```

# Check n is power of 2

```python
"""
2 = 010
1 = 001
2 & 1 = 000 => 0

4 = 100
3 = 011
4 & 3 = 000 => 0

16 = 10000
15 = 01111
16 & 15 = 00000 => 0

from above we can say that 
if n & (n-1) == 0 => n is power of 2
"""
```

```python
def power(n):
    if n < 0:
        return False
    return n & (n-1) == 0

print(power(20))
print(power(32))
```

# Count Number of Set bits

```python
# our n & (n-1) will remove the last set bit from n
# so we can update our count by 1 every time we remove the last set bit from n till n become 0

def countSetBits(n):
    count = 0
    while  n:
        n = n & (n-1)
        count += 1
    return count 


n = 20
print(bin(n)[2:])
print(countSetBits(n))
```
